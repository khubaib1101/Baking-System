```python
# Final Project: Mini Banking System

def check_balance(balance):

    print("Your current balance is-->",balance,'Rps')




def deposit(balance,amount):


      balance = balance + amount
      print("You deposited:",amount,"and yout current balance is:",balance)


def withdraw(balance,amount):


    if amount>0 and amount<=balance:
      balance = balance - amount

      if balance>=1000:
        print("Transaction Completed:",balance)

      else:
        print("Transaction Rejected")
        print("Reason: Balance is less than 1000")


def loan_eligibility(age,salary,creditScore):

    if age>=21:
      if salary>=50000:
        if creditScore>=650:
          print("You are Eligible for Loan")
        else:
          print("You are not Eligible")
          print("Reason:Credit Score is less than 650")
      else:
        print("You are not Eligible")
        print("Your Salary is less than 50,000")
    else:
      print("You are not Eligible")
      print("Reason:Your age is less than 21")







print("1.Check Balance")
print("2.Deposit")
print("3.Withdraw")
print("4.Loan Eligibility")
print("5.Exit")

balance = 50000
choice = int(input("Enter Your Choice:"))
if choice == 1:
  check_balance(balance)
elif choice == 2:
  amount = int(input("Enter your Amount:"))
  deposit(balance,amount)
elif choice == 3:
  amount = int(input("Enter your Amount:"))
  withdraw(balance,amount)
elif choice == 4:
  age = int(input("Enter Your Age:"))
  salary = int(input("Enter your Salary:"))
  creditScore = int(input("Enter your Credit Score:"))
  loan_eligibility(age,salary,creditScore)
elif choice == 5:
  print("You Exited The Program")
```
